#### 실제 코드

```python
import time
import random
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
```

```python
chrome_options = Options()
chrome_options.add_argument("--incognito")
chrome_options.add_argument("--disable-blink-features=AutomationControlled")
chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
chrome_options.add_experimental_option('useAutomationExtension', False)
```

```python
categories = [
    "블라우스", "셔츠", "니트", "원피스", "투피스", "점프수트", "바지", "치마",
    "티셔츠", "집업", "가디건", "패딩", "바람막이", "자켓", "점퍼", "코트",
    "트렌치코트", "무스탕", "야상", "정장 바지", "정장 세트", "정장 자켓",
    "정장 치마", "트레이닝 상의", "트레이닝 세트", "트레이닝 하의",
    "레깅스", "맨투맨", "후드", "나시", "조끼 상의", "조끼 아우터", "상하의 세트"
]

prompt_list = []

for category in categories:
    particle = "을" if category in ['집업', '가디건', '패딩', '자켓', '무스탕', '야상', '정장 자켓', '맨투맨'] else "를"
    
    prompt = (f"사람들이 너에게 {category}{particle} 찾아달라고 할 때 던지는 질문을 30가지 출력해줘. 소재, 기장, 계절, 색감 등을 자유롭게 변형하여 30가지 작성해야 해. 답변 시 필요없는 설명구는 다 빼고, 오직 질문만 출력해. 답변 말투는 정말 사람들이 너에게 던지는 질문의 어투를 사용하되, 첫 질문 15가지에서는 신조어를 정제하고, 나머지 질문 15가지에서는 신조어를 포함해서 작성해줘."
    )
    
    prompt_list.append({"question": prompt})

import json
print(json.dumps(prompt_list, ensure_ascii=False, indent=2))
```

```python
collected_data = []
driver = webdriver.Chrome(options=chrome_options)

try:
    # 최초 페이지 접속
    driver.get("https://gemini.google.com/app")
    
    for index, item in enumerate(prompt_list):
        question = item["question"]
        
        print(f"\n[{index + 1}/{len(prompt_list)}] [{index}]번째 질문 전송 중: {question}")
        
        wait = WebDriverWait(driver, 15)
        
        try:
            input_selector = (By.CSS_SELECTOR, 'div[contenteditable="true"]')
            input_element = wait.until(EC.element_to_be_clickable(input_selector))
            
            input_element.click()
            time.sleep(random.uniform(0.5, 1.0))
            
            for char in question:
                input_element.send_keys(char)
                time.sleep(random.uniform(0.05, 0.15))
                
            input_element.send_keys(Keys.ENTER)
            print("질문 전송 완료. 답변 생성 대기 중...")
            
            response_selector = (By.CSS_SELECTOR, 'div[id*="model-response-message"]')
            wait.until(EC.presence_of_element_located(response_selector))
            
            time.sleep(20) 
            
            p_tags_selector = 'div[id*="model-response-message"] li p'
            p_elements = driver.find_elements(By.CSS_SELECTOR, p_tags_selector)
            
            answers_list = [p.text.strip() for p in p_elements if p.text.strip()]
            
            if not answers_list:
                response_element = driver.find_element(*response_selector)
                answers_list = [line.strip() for line in response_element.text.split('\n') if line.strip()]
                
            print(f"▼ 답변 수집 완료 (총 {len(answers_list)}개 항목)")
            
            collected_data.append({
                "질문": question,
                "답변": answers_list
            })
            
        except Exception as e:
            print(f"❌ [{index + 1}번째] 처리 중 에러 발생: {e}")
            collected_data.append({
                "질문": question,
                "답변": ["ERROR: 답변 수집 실패"]
            })
            
        print("세션 초기화를 위해 새로고침 중...")
        driver.refresh()
        time.sleep(random.uniform(7.0, 12.0))

finally:
    driver.quit()
    print("\n🎉 크롤링 프로세스 종료.")


if collected_data:
    df = pd.DataFrame(collected_data)
    df_expanded = df.explode('답변').reset_index(drop=True)

    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    
    print("\n================ 수집 완료 데이터프레임 ================")
    print(df_expanded.head(20)) # 상위 20개 행만 우선 출력해 확인
    print("======================================================")

    file_name = "gemini_collected_data.csv"
    df_expanded.to_csv(file_name, index=False, encoding="utf-8-sig")
    print(f"💾 데이터가 분할되어 '{file_name}' 파일로 성공적으로 저장되었습니다!")
else:
    print("수집된 데이터가 없습니다.")
```
